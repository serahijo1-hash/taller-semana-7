```python
# 1. Función sin parámetros ni retorno (Reto 1)

def mostrar_mensaje():
    print("Procesando Dataset de Trabajo")

# Invocación de la función
mostrar_mensaje()
```

```python


# 2. Función con parámetro (Reto 2)
def saludar(nombre):
    print(f"Hola estudiante: {nombre}")

# 3. Función con retorno simple (Reto 3 y 5)
def clasificar_evento(evento):
    """Clasifica el evento de forma sencilla con if-else"""
    if evento == "Intento de Suicidio":
        return "CRITICO"
    else:
        return "ALERTA"

# 4. Función para mostrar el menú (Reto 9)
def mostrar_menu():
    print("\n--- MENÚ PRINCIPAL ---")
    print("1. Mostrar mensaje de bienvenida")
    print("2. Probar clasificación")
    print("3. Salir")

# 5. Función para leer la opción del usuario (Reto 10)
def leer_opcion():
    opcion = input("Digite una opción: ")
    return opcion

# 6. Función principal (main)
def main():
    # ponemos la función del Reto 1
    mostrar_mensaje()

    # ponemos la función del Reto 2
    saludar("Carlos")

    # Ciclo básico para el menú
    continuar = True
    while continuar:
        mostrar_menu()
        opcion = leer_opcion()

        if opcion == "1":
            print("¡El Dataset de Trabajo está listo y funcionando!")
        elif opcion == "2":
            # Probamos la función con el retorno
            resultado = clasificar_evento("Intento de Suicidio")
            print(f"La clasificación del evento es: {resultado}")
        elif opcion == "3":
            print("Saliendo del programa...")
            continuar = False
        else:
            print("Opción no válida, intente de nuevo.")

# Ejecutamos lo principal del programa
if __name__ == "__main__":
    main()
```
